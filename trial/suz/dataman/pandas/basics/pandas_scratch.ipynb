### Row filters
'Where' clause in SQL

```python
# 1 row, with name = Alice
df[ df['Name'] == 'Alice' ]
```

```python
# Filter to exclude Alice
df[ df['Name'] != 'Alice' ]
```

```python
# filter all rows, who are working in IT
df[ df['Dept'] == 'IT' ]
```

```python
# filter 1 row, with max salary
max = df[['Salary']].max().values[0]
df[ df['Salary'] == max ]
```

Above works, but it has a few minor inefficiencies: it shadows Python's built-in max function, extracts the value using .values[0] unnecessarily, and creates a DataFrame subset [['Salary']] instead of a faster Series ['Salary'].

Here are the best ways to optimize this depending on how you want to handle ties (multiple rows having the same max salary):

```python
df[ df['Salary'] == df['Salary'].max() ]
```

```python
# All except the heighest salary
df[ df['Salary'] != df['Salary'].max() ]
```

```python
# filter all rows, who are working in IT
# Add add / aggregate all their salaries
'''
 SELECT SUM(SALARY)
 FROM EMPLOYEE
 WHERE DEPT IS 'IT'
'''
df[ df['Dept'] == 'IT' ][['Salary']].sum()
```

```python
'''
 SELECT AVG(AGE)
 FROM EMPLOYEE
 WHERE DEPT IS 'IT'
'''
df[ df['Dept'] == 'IT' ][['Age']].mean()
```

### Group By and aggregation

```python
# Group BY Department, and count employees per department
print(df.groupby(by='Dept')[['EmpID']].count()['EmpID'].sum())
display(df.groupby(by='Dept')[['EmpID']].count())
```

```python
df.groupby(by='Age')[['Dept']].count()
```

```python
display(df.groupby(by='Age')[['Salary']].sum())
display(df.groupby(by='Dept')[['Salary']].sum())
```

```python
# Don't group-by primary-key, or unique-key ==> No grouping advantage
display(df.groupby(by='EmpID')[['Salary']].sum())
```

```python
df.groupby(by='Age')[['Dept']].value_counts()
```

```python
# Same as above, but results into a series
df.groupby(by='Dept')['Dept'].count()

# Can also be done using:
# df[['Dept']].value_counts()
```

```python
# Group by Dept, sum of salaries by department.
df.groupby(by='Dept', group_keys=['Dept'])[['Salary']].sum()
```

#### Group By, with filters (SQL 'having')

```python
# Group by Dept, sum of salaries by department, having aggregated-salary > 62000
df.groupby(by='Dept', group_keys=['Dept'])[['Salary']].sum().query('Salary > 220000.0')
```

```python
# Alt:
grouped_series = df.groupby(by='Dept', group_keys=['Dept'])[['Salary']].sum()
grouped_series[ grouped_series['Salary'] > 220000.0 ]
```

```python
# Find all departments where the number of employees in the department is more than 3
grouped_count_df = df.groupby(by='Dept')[['Dept']].count()
grouped_filtered_df = grouped_count_df.query('Dept > 2')
display(grouped_filtered_df)
print("Keys: %s" % list(grouped_filtered_df.keys()))
print("Values: %s" % grouped_filtered_df.values)
```

### Joins (Inner)

#### Order Table

```python
data_s2 = {
    'EmpID': [101, 102, 103, 104, 101, 102, 103, 101 ],
    'OrdId': ['ORD1', 'ORD2', 'ORD3', 'ORD4', 'ORD5', 'ORD6', 'ORD7', 'ORD8'],
    'Amount': [100, 150, 50, 600, 100, 80, 20, 50],
}
df2 = pd.DataFrame(data_s2)
display(df2)
```

```python
display(df2.groupby(by='EmpID')[['OrdId']].count())
display(df2.groupby(by='EmpID')[['Amount']].sum().query('Amount > 500'))
```

```python
pd.merge(df, df2,
         # Optional
         on='EmpID',
         # By default it is inner
         #how='inner'
         )
```

```python
pd.merge(df, df2,
         # Optional
         on='EmpID',
         how='left'
         )
```

```python
pd.merge(df, df2,
         # Optional
         on='EmpID',
         how='right'
         )
```
