## Scenario 1: Data Initialization & Inspection
**Objective:** Load raw tabular data into a Pandas DataFrame and inspect its structure.

Before we can analyze data, we have to get it into a DataFrame. A DataFrame is essentially a programmatic Excel spreadsheet. Here, we'll create one from scratch using a Python dictionary, and then use built-in methods to understand what we're looking at.

* **`.head()`**: Shows the first 5 rows.
* **`.info()`**: Shows column data types and counts missing values.
* **`.describe()`**: Gives you instant math (mean, min, max) on numerical columns.

## DataFrame and Series
Analogy:
* **DataFrame** is like a Matrix, while,
* **Series** is like a Vector

```python
# 1. Initialize DataFrame from a dictionary
data_s1 = {
    'EmpID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Evan'],
    'Dept': ['IT', 'HR', 'IT', 'Marketing', 'IT'],
    'Salary': [75000.0, 60000.0, 82000.0, 65000.0, 90000.0]
}
df_employees = pd.DataFrame(data_s1)

# 2. Inspect the data
print("--- FIRST 3 ROWS ---")
display(df_employees.head(3)) # using display() instead of print() looks prettier in Jupyter

print("\n--- DATAFRAME INFO ---")
df_employees.info()

print("\n--- STATISTICAL SUMMARY ---")
display(df_employees.describe())

# PRACTICE: Try changing df_employees.head(3) to df_employees.tail(2) to see the bottom rows!
```

### Slicing and Dicing DataFrame

##### Slicing

```python
# Basic operations / attributes of DF
print(df_employees.keys())

# Retrieve all columns
display(df_employees[df_employees.keys()])
print()

# Retrieve selected columns
print(f"Display columns: {df_employees.keys()[1:-1]}")
display(df_employees[df_employees.keys()[1:-1]])
print("ALT:")
# Retrieve selected columns - by specific col-names
## Each Slicing creates a new DF
print(f"Display columns: [Name, Salary]")
display(df_employees[['Name', 'Salary']])
print(type(df_employees[['Name', 'Salary']]))
print()

# Does not work!
# display(df_employees[1:-1][])
# display(type(df_employees[[0, 1]]))
```

##### Dicing

```python
# Dicing: (row selection)
## Select alt. rows starting row_index=0, until/before -1
display(df_employees[0:-1:2])

# Each slicing and/or dicing creates a new DF,
# so the indices (corresponding to each dimension/axis) can be
# appended in any order, leading to the same result.
display(df_employees[0:-1:2][['Name', 'Salary']])
display(df_employees[['Name', 'Salary']][0:-1:2])
```

### Loc and iLoc

```python
## iLoc (select row-column rectangle - using multiple axes with integer indices)
display(df_employees.iloc[0:, 1:-1])
display(df_employees.iloc[0:, [1, 1]])

print()
## Loc is used for specifying non-integer indices (or keys, or col-names)
display(df_employees.loc[0:, ['Name', 'Salary']])
print()
```

#### Basic Aggregation and Sort Operations
- count - counts non-null (or non-NA) entries, either
    - per row (collapsing all entries per row) or axis=1
        - Results into column-vector
    - per col (collapsing all entries per column) or axis=0 [default]
        - Results into row-vector
- value_counts - Counts unique values (tuples)

```python
# Count
## display(df_employees.count()) # default axis = 0

## Results into row-vector, counting non-nulls per column
print(df_employees.count(axis=0)) # This unfortunately
                                  # prints row-wise as a col-vector
## Making it row-vector (as the aggregation actually happens
# by collapsing cols)
display(pd.DataFrame(df_employees.count(axis=0)).transpose())
print()
## Results into column-vector, counting non-nulls entries in the record, per row
display(df_employees.count(axis=1))
print()
```

```python
# Value_Counts

## Unique entries (entire row compared)
display(df_employees.value_counts())
print()
## Unique entries (With few columns)
display(df_employees[['Dept']].value_counts())
```

```python
# 'sum' uses + operator (concat for str, while add for int/float dType)
## Advisable to sum column-wise (resulting into row-vector)
display(df_employees.sum(axis=0))
print()

## This may not work, as + is not defined for: str + int
# display(df_employees.sum(axis=1))

## Practical use, sum all salaries
display(df_employees[['Salary']].sum(axis=0))
```

```python
## Practical use, sum all columns (including 'Salary') grouped-by Dept
display(df_employees.groupby(by=['Dept']).sum())

## Practical use, count all entries grouped-by Dept
display(df_employees.groupby(by=['Dept']).count())

## Practical Use (Group By):
## Multiple aggregate functions, for a given numeric column
display(df_employees.groupby(by=['Dept'])['Salary'].agg(['sum', 'mean', 'count', 'min', 'max' ]))

display(df_employees.groupby(by=['Dept'])['Name'].agg(['sum', 'mean', 'count']))
```

## Scenario 2: Data Filtering (Boolean Indexing)
**Objective:** Extract a specific subset of data based on logical conditions.

You rarely need to look at *all* your data at once. You usually want to ask questions like: "Show me all IT employees who make more than $70,000." We do this using Boolean Indexing.

**Rule of Thumb:** 
Wrap each condition in parentheses `()` and use `&` for AND, and `|` for OR.

```python
# Create our filter condition
# Condition 1: Department is IT
# Condition 2: Salary is greater than 70000
mask = (df_employees['Dept'] == 'IT') & (df_employees['Salary'] > 70000)

# Apply the mask to the dataframe
high_earning_it = df_employees[mask]

print("High Earning IT Employees:")
display(high_earning_it)

# Select only specific columns from that filtered result
names_and_salaries = high_earning_it[['Name', 'Salary']]
display(names_and_salaries)

# PRACTICE: Create a new filter that finds anyone in 'HR' OR anyone making exactly 65000.0
```

## Scenario 3: Data Cleaning (Handling Missing Values)
**Objective:** Identify and fix missing values (`NaN`) so they don't break our math later.

Real-world data is messy. Pandas represents blank or missing data as `NaN` (Not a Number). We can handle this by either dropping the rows completely (`.dropna()`) or filling them with a placeholder/average (`.fillna()`).

```python
# Create dirty data with missing values
dirty_data = pd.DataFrame({
    'ProductID': [1, 2, 3, 4],
    'Price': [25.50, np.nan, 15.00, np.nan],
    'Category': ['Electronics', 'Clothing', None, 'Electronics']
})

print("--- ORIGINAL DIRTY DATA ---")
display(dirty_data)

print("\n--- MISSING VALUE COUNTS ---")
print(dirty_data.isnull().sum())

# Clean the data (Imputation)
clean_df = dirty_data.copy()

# 1. Fill missing prices with the average (mean) price
avg_price = clean_df['Price'].mean()
clean_df['Price'] = clean_df['Price'].fillna(avg_price)

# 2. Fill missing categories with the word 'Unknown'
clean_df['Category'] = clean_df['Category'].fillna('Unknown')

print("\n--- CLEANED DATA ---")
display(clean_df)

# PRACTICE: Try using clean_df.dropna() instead of fillna() and see how the dataframe changes.
```

## Scenario 4: Grouping & Aggregation
**Objective:** Collapse the data based on categories to calculate summary statistics.

Think of this like a Pivot Table. If you have a list of sales, you probably want to know the *total* sales per region, or the *average* sales per store. We use `.groupby()` followed by an aggregation function like `.sum()`, `.mean()`, or `.count()`.

```python
sales_df = pd.DataFrame({
    'Region': ['North', 'North', 'South', 'South', 'East'],
    'Store': ['A', 'B', 'C', 'D', 'E'],
    'Revenue': [5000, 7000, 4000, 6000, 3000]
})

display(sales_df)

# Group by the 'Region' column
grouped = sales_df.groupby('Region')

# Calculate the sum of Revenue for each Region
total_revenue = grouped['Revenue'].sum()

print("--- TOTAL REVENUE BY REGION ---")
display(total_revenue)

# Advanced: Calculate multiple metrics at once!
multi_summary = grouped['Revenue'].agg(['sum', 'mean', 'count'])
print("\n--- DETAILED REGIONAL SUMMARY ---")
display(multi_summary)

# PRACTICE: Try grouping by 'Store' instead of 'Region'.
```

## Scenario 5: Merging DataFrames (SQL-style Joins)
**Objective:** Combine relational data from two distinct tables.

Often, your data is split across different files (e.g., a file for User details, and a file for their Orders). We use `pd.merge()` to stick them together based on a shared column (a key).

```python
users = pd.DataFrame({
    'UserID': [1, 2, 3], 
    'Name': ['John', 'Jane', 'Zack']
})

orders = pd.DataFrame({
    'OrderID': [101, 102, 103, 104],
    'UserID': [1, 1, 2, 'invalid_id'],
    'Amount': [250.0, 50.0, 120.0, 3300.0]
}) # Note: Zack (UserID 3) has no orders.

## on = < foreign_key(s) >

# INNER JOIN: Only keeps users who HAVE orders
inner_merge = pd.merge(users, orders, on='UserID', how='inner')
print("--- INNER JOIN ---")
display(inner_merge)

# LEFT JOIN: Keeps ALL users, even if they have no orders (fills with NaN)
left_merge = pd.merge(users, orders, on='UserID', how='left')
print("\n--- LEFT (OUTER) JOIN ---")
display(left_merge)

# LEFT JOIN: Keeps ALL users, even if they have no orders (fills with NaN)
left_merge = pd.merge(users, orders, on='UserID', how='right')
print("\n--- RIGHT (OUTER) JOIN ---")
display(left_merge)

# PRACTICE: Look at the LEFT JOIN result. Try chaining .fillna(0) to replace Zack's NaN order amount with 0.
```

## Scenario 6: Time Series & Resampling
**Objective:** Handle dates and times cleanly, and group data by time periods (e.g., converting hourly data into daily data).

Pandas was originally built for financial data, so it handles dates incredibly well. Once you set a `datetime` column as your index, you can use `.resample()` to easily group by Days ('D'), Months ('M'), or Years ('Y').

```python
# Generate 48 hours of dummy sensor data
dates = pd.date_range(start='2026-01-01', periods=48, freq='h') 
sensor_df = pd.DataFrame({
    'Timestamp': dates,
    'Temp': np.random.uniform(15, 25, size=48)
})

# 1. Ensure column is datetime type (it already is here, but good practice!)
sensor_df['Timestamp'] = pd.to_datetime(sensor_df['Timestamp'])

# 2. Set the datetime column as the index
sensor_df.set_index('Timestamp', inplace=True)

print("--- HOURLY DATA (First 4 hours) ---")
display(sensor_df.head(4))

# 3. Resample the data to Daily ('D') averages
daily_average = sensor_df.resample('D').mean()

print("\n--- DAILY AVERAGE ---")
display(daily_average)

# PRACTICE: Try resampling by 12 hours ('12h') instead of 'D'.
```

## Scenario 7: Applying Custom Functions (`.apply`)
**Objective:** Run a custom Python function over every row or column in a DataFrame.

Sometimes the built-in math isn't enough. If you need complex logic (like categorizing ages into "Child", "Adult", "Senior"), you write a normal Python function and `.apply()` it to the column.

```python
age_df = pd.DataFrame({'Name': ['Tom', 'Sue', 'Ali'], 'Age': [12, 34, 68]})

# Define our custom logic
def categorize_age(age):
    if age < 18:
        return 'Child'
    elif age < 65:
        return 'Adult'
    else:
        return 'Senior'

# Apply the function to the 'Age' column to create a new 'Category' column
age_df['Category'] = age_df['Age'].apply(categorize_age)

display(age_df)

# PRACTICE: Write a lambda function inside .apply() to add 5 years to everyone's age.
# Example: age_df['Age_in_5_years'] = age_df['Age'].apply(lambda x: x + 5)
```

## Scenario 8: Pivot Tables
**Objective:** Reshape data to show relationships between two categorical variables.

Pivot tables in Pandas work exactly like they do in Excel. They allow you to look at your data from multiple angles simultaneously.

```python
# Let's say we have car sales data
car_sales = pd.DataFrame({
    'Make': ['Toyota', 'Ford', 'Toyota', 'Ford', 'Toyota', 'Ford'],
    'Type': ['Sedan', 'Truck', 'Truck', 'Sedan', 'Sedan', 'Truck'],
    'Price': [25000, 40000, 35000, 22000, 26000, 42000]
})

# We want to see the Average Price, broken down by Make (Rows) and Type (Columns)
pivot = pd.pivot_table(
    car_sales, 
    values='Price', 
    index='Make', 
    columns='Type', 
    aggfunc='mean',
    fill_value=0 # Fills empty combinations with 0
)

display(pivot)

# PRACTICE: Change the aggfunc to 'sum' to see total revenue instead of average price!
```

## Scenario 9: String Operations (`.str`)
**Objective:** Clean and manipulate text data quickly.

If you have a column of text, you can use the `.str` accessor to apply standard Python string methods (like `.lower()`, `.replace()`, or `.split()`) to the entire column at once.

```python
text_df = pd.DataFrame({
    'Raw_Names': ['  JOHN doe ', 'JANE SMITH', '  albert EINSTEIN  ']
})

# 1. Strip whitespace
text_df['Clean'] = text_df['Raw_Names'].str.strip()

# 2. Make everything Title Case
text_df['Clean'] = text_df['Clean'].str.title()

# 3. Split the name into First and Last
# expand=True turns the split list into separate DataFrame columns
text_df[['First_Name', 'Last_Name']] = text_df['Clean'].str.split(' ', expand=True)

display(text_df)

# PRACTICE: Try to create a column containing the length of the names using .str.len()
```
