### Question 2: Relational Merging and Averaging
Calculate the average salary of employees who have placed at least one team expense order,
grouped by the Department Building.

_Hint: Dept_code is not 1:1 with Building_

```python
display(df_ord[df_ord['isTeamExpense'] == True].merge(df_emp, how='inner', on='EmpId'))
df_ord[df_ord['isTeamExpense'] == True].merge(df_emp, how='inner', on='EmpId').\
    groupby(by='Dept_Code')[['Salary']].mean()
```

```python
#Avg
print("Incorrect Avg1: %d" % ((95000*2+80000)/3))
print("Incorrect Avg2: %d" % ((95000+80000)/2))
```

```python
# Filter for team expenses and get unique employees
team_exp_emps = df_ord[df_ord['isTeamExpense']]['EmpId'].unique()
display(team_exp_emps)

# Filter employee dataframe
emp_team = df_emp[df_emp['EmpId'].isin(team_exp_emps)]
display(emp_team)

# Merge with departments to get building, then group by building
merged_bldg = emp_team.merge(df_dept, on='Dept_Code')
display(merged_bldg)
```

```python
# More than one dept can be allocated the same bldg (e.g. 'Alpha')
print("Correct Avg (Bldg-wise): %d" % ((95000*2+80000)/3))
```

```python
# Filter for team expenses and get unique employees
team_exp_emps = df_ord[df_ord['isTeamExpense']]['EmpId'].unique()

# Filter employee dataframe
emp_team = df_emp[df_emp['EmpId'].isin(team_exp_emps)]

# Merge with departments to get building, then group by building
merged_bldg = emp_team.merge(df_dept, on='Dept_Code')
avg_salary_bldg = merged_bldg.groupby('Building')['Salary'].mean()
print(avg_salary_bldg)
```
