## Designing an AND Gate Perceptron using tensor Keras Layers

This notebook demonstrates how to build a single-neuron Perceptron to model the logical `AND` function. The architecture is derived from computational models inspired by biological neurons, utilizing inputs, weights, a bias, and an activation function[cite: 1].

The `AND` gate takes two binary inputs ($x_1$ and $x_2$) and outputs `1` only if both inputs are `1`. Otherwise, it outputs `0`.

```python
import tensorflow as tf
import numpy as np
```

## 1. Defining the Dataset

We will represent the data features as inputs ($x_i$)[cite: 1]. The `AND` function has four possible combinations:
*   `[0, 0]` $\rightarrow$ `0`
*   `[0, 1]` $\rightarrow$ `0`
*   `[1, 0]` $\rightarrow$ `0`
*   `[1, 1]` $\rightarrow$ `1`

```python
# Define the inputs (x1, x2)
X = np.array([[0, 0], 
              [0, 1], 
              [1, 0], 
              [1, 1]], dtype=float)

# Define the expected outputs (y)
y = np.array([[0], 
              [0], 
              [0], 
              [1]], dtype=float)
```

## 2. Anatomy of a Single Neuron (Perceptron)

We will construct a simple model with a single layer containing one neuron. Under the hood, this neuron computes the weighted sum (logits/pre-activation) using the formula: $z = \sum (w_i x_i) + b$.

To introduce non-linearity, we apply a Sigmoid activation function, which acts as a squashing function to convert the raw sum into a probability between 0 and 1.

```python
# Build the model
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_dim=2, activation='sigmoid')
])

model.summary()
```

## 3. Training the Network: Optimization & Learning

To train the model, we need a way of measuring *how wrong* the model is, which is handled by the loss function[cite: 1]. We will use Binary Cross-Entropy because this is a binary classification problem. 

We also need an optimizer to act as the mathematical compass[cite: 1]. We will use the Adam optimizer to control the learning rate step size and perform gradient descent[cite: 1].

```python
# Compile the model
model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.1),
              loss='binary_crossentropy',
              metrics=['accuracy'])
```

## 4. Model Training (Forward & Backpropagation)

During the `fit` process, data flows sequentially from input to output via forward propagation[cite: 1]. The model then computes the loss and applies backpropagation—using the Chain Rule of Calculus recursively—to adjust the weights using the computed gradients[cite: 1].

```python
# Train the perceptron
history = model.fit(X, y, epochs=500, verbose=0)
print("Training complete.")
```

## 5. Evaluation & Final Predictions

Now that the network has been optimized, we pass our inputs back through the model. The output will be continuous probability values, which we threshold at `0.5` to yield the final boolean `AND` gate results.

```python
# Predict using the trained model
predictions = model.predict(X)

print("\n--- AND Gate Predictions ---")
for i, (input_val, pred) in enumerate(zip(X, predictions)):
    final_output = 1 if pred[0] >= 0.5 else 0
    print(f"Input: {int(input_val[0])}, {int(input_val[1])} | Raw Probability: {pred[0]:.4f} | Final Output: {final_output}")
```
